### 0/1 Knapsack Problem

Classic DP problem where you’re given:
- `weights[]`: weights of `n` items
- `values[]`: value of each item
- `capacity`: max weight you can carry

#### Goal:
Maximize the total value without exceeding capacity. You **cannot take fractions** or reuse items.

#### State:
- `dp[i][w]` = max value using first `i` items with total weight ≤ `w`

#### Time Complexity:
- O(n·W) where n = items, W = capacity


```python
def knapsack(weights, values, capacity):
    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(capacity + 1):
            if weights[i - 1] <= w:
                dp[i][w] = max(dp[i - 1][w], dp[i - 1][w - weights[i - 1]] + values[i - 1])
            else:
                dp[i][w] = dp[i - 1][w]

    return dp[n][capacity]
```
