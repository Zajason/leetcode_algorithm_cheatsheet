### Backtracking

Backtracking is a recursive algorithm used to explore **all possible combinations or permutations**. It builds candidates for solutions step-by-step and abandons them if they fail to meet the constraints ("backtrack").

#### Key Steps:
1. Choose: Add a new candidate to the current path.
2. Explore: Recursively explore further candidates.
3. Un-choose (Backtrack): Remove the last choice and try another.

#### Use Cases:
- Generating permutations or combinations
- Solving constraint problems (e.g. Sudoku, N-Queens)
- Subset generation, pathfinding

#### Advantages:
- Clean and elegant recursive structure
- Solves problems that require exhaustive search with pruning

#### Drawbacks:
- Can be exponential in time if not pruned correctly


```python
def permute(nums):
    res = []

    def backtrack(path, remaining):
        if not remaining:
            res.append(path[:])
            return
        for i in range(len(remaining)):
            backtrack(path + [remaining[i]], remaining[:i] + remaining[i+1:])

    backtrack([], nums)
    return res
```
