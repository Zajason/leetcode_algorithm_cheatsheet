### Dynamic Programming – Tabulation (Bottom-Up)

Tabulation builds up the solution **iteratively** from the smallest subproblem to the final goal using an array or table.

#### Key Steps:
1. Create a DP array of size `n+1`.
2. Set base cases (e.g. `dp[0]`, `dp[1]`).
3. Use a loop to fill in `dp[i]` using previous values.

#### Use Cases:
- More space/time efficient than recursion + memo in some cases
- When you need to build full DP table (e.g., 0/1 Knapsack, LIS)

#### Time Complexity:
- Typically O(n) or O(n·k)


```python
def climb_stairs_tab(n):
    if n <= 2:
        return n
    dp = [0] * (n + 1)
    dp[1], dp[2] = 1, 2
    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]
```
