### Dynamic Programming – Memoization (Top-Down)

Memoization is a top-down dynamic programming approach that stores results of expensive function calls and returns the cached result when the same inputs occur again.

#### Key Steps:
1. Define a recursive function that solves the problem.
2. Store the results of subproblems in a dictionary (`memo`) or array.
3. Reuse stored results instead of recomputing them.

#### Use Cases:
- Problems with **overlapping subproblems**
- Recursive problems where brute force is exponential
- Fibonacci, Climbing Stairs, Coin Change, etc.

#### Time Complexity:
- Reduced from exponential to polynomial


```python
def climb_stairs(n, memo={}):
    if n <= 2:
        return n
    if n not in memo:
        memo[n] = climb_stairs(n - 1, memo) + climb_stairs(n - 2, memo)
    return memo[n]
```
