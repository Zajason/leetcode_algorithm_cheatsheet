### Min/Max Heap (Priority Queue)

A heap is a specialized tree-based structure where the parent node is either:
- ≤ (min-heap)
- ≥ (max-heap) its children

In Python, `heapq` provides a min-heap implementation.

#### Use Cases:
- Find k largest/smallest elements
- Priority scheduling
- Dijkstra’s algorithm

#### Time Complexity:
- Insertion: O(log n)
- Access min/max: O(1)
- Pop min/max: O(log n)


```python
import heapq

# Min-heap example
def k_smallest(nums, k):
    heapq.heapify(nums)
    return [heapq.heappop(nums) for _ in range(k)]

# Max-heap using negation
def k_largest(nums, k):
    max_heap = [-num for num in nums]
    heapq.heapify(max_heap)
    return [-heapq.heappop(max_heap) for _ in range(k)]
```
