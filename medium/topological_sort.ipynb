### Topological Sort

Topological sort is an algorithm for **ordering nodes in a directed acyclic graph (DAG)** such that for every directed edge `u → v`, node `u` comes before `v`.

#### Kahn's Algorithm (BFS-based):
1. Calculate in-degrees of all nodes.
2. Add all nodes with in-degree 0 to the queue.
3. Pop from queue and reduce in-degrees of neighbors.
4. If a neighbor’s in-degree drops to 0, add to queue.

#### Use Cases:
- Course scheduling
- Task dependency resolution
- Package installation order

#### Time Complexity:
- O(V + E)


```python
def topological_sort(n, edges):
    from collections import defaultdict, deque

    graph = defaultdict(list)
    in_degree = [0] * n

    for u, v in edges:
        graph[u].append(v)
        in_degree[v] += 1

    queue = deque([i for i in range(n) if in_degree[i] == 0])
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)

        for neighbor in graph[node]:
            in_degree[neighbor] -= 1
            if in_degree[neighbor] == 0:
                queue.append(neighbor)

    return order if len(order) == n else []  # return empty if cycle
```
