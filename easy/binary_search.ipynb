### Binary Search

Binary Search is a **divide and conquer** algorithm used to search a **sorted** array in O(log n) time by repeatedly dividing the search space in half.

#### Key Steps:
1. Set `left = 0` and `right = n - 1`.
2. Find mid = (left + right) // 2.
3. If mid is the target, return it.
4. If target < mid, move right = mid - 1.
5. If target > mid, move left = mid + 1.

#### Use Cases:
- Searching in sorted arrays
- Finding boundaries or breakpoints (e.g., first bad version, smallest/largest element meeting criteria)

#### Advantages:
- Very fast for large sorted data sets.


```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
```
