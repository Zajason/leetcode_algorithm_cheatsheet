### Fast & Slow Pointers (Tortoise and Hare)

A two-pointer approach used to detect cycles in data structures like linked lists or recurring numerical sequences.

#### Key Steps:
1. Start both pointers at the head or start index.
2. Move `slow` one step at a time, `fast` two steps.
3. If they ever meet, a cycle exists.
4. To find the start of the cycle, reset one pointer and move both one step at a time.

#### Use Cases:
- Detecting cycles in linked lists
- Finding repeated numbers
- Cycle length problems

#### Advantages:
- O(n) time, O(1) space


```python
def has_cycle(head):
    slow = fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

        if slow == fast:
            return True

    return False
```
