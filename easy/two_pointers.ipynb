### Two Pointers Technique

The two pointers technique is used when dealing with **sorted arrays**, **linked lists**, or **palindromic problems**. It involves using two indices that move toward each other or in the same direction to find a solution.

#### Key Steps:
1. Initialize two pointers at either end (or start) of the list.
2. Move them based on some condition until they meet or cross.
3. Return the answer based on problem constraints (e.g., indices, count, boolean).

#### Use Cases:
- Finding pairs that sum to a target in a sorted array.
- Removing duplicates.
- Partitioning an array.

#### Advantages:
- Avoids nested loops, reducing time complexity to O(n).


```python
def two_sum_sorted(nums, target):
    left, right = 0, len(nums) - 1

    while left < right:
        curr_sum = nums[left] + nums[right]
        if curr_sum == target:
            return [left, right]
        elif curr_sum < target:
            left += 1
        else:
            right -= 1
```
