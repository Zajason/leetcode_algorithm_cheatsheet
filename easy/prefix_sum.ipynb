### Prefix Sum

Prefix Sum is a technique used to optimize **range sum queries** in constant time after a one-time linear preprocessing step.

#### Key Steps:
1. Create a prefix array where prefix[i] = sum of elements from 0 to i-1.
2. For any range [i, j], the sum is `prefix[j+1] - prefix[i]`.

#### Use Cases:
- Sum of subarrays in O(1)
- Finding subarrays with a given sum
- Optimizing nested loops in brute-force sum problems

#### Advantages:
- Improves efficiency of range sum calculations from O(n) to O(1)


```python
def build_prefix_sum(nums):
    prefix = [0]
    for num in nums:
        prefix.append(prefix[-1] + num)
    return prefix

def range_sum(prefix, i, j):
    return prefix[j + 1] - prefix[i]
```
